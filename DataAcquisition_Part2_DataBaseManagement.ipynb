Database Management:

To manage the lumbar spine MRI dataset, a simple relational database structure is designed to organize metadata associated with each MRI image. 
Since the dataset consists of preprocessed and labeled MRI images, the primary structured entities include image information and diagnostic labels.
The database consists of an Images table, which stores metadata for each MRI image such as a unique image identifier, file path, image dimensions. 
Each image is associated with exactly one diagnostic label. A Labels table is used to store the possible classification categories (herniated disc, thecal sac involvement, and no stenosis). 

There are two tables created: 
- Train.csv
- Test.csv

The database structure can be represented as:

Images (image_id, image_link, image_label)

Basically, the project works with unstructured data in the form of MRI image files. 
These image files are not stored directly in the database due to their large size.
Instead, they are stored in the file system, and their locations are referenced through links stored in the Images table.

# Train dataset to csv

```python
import pandas as pd
from huggingface_hub import list_repo_files
from urllib.parse import quote

DATASET_ID = "hitikaghanani/project_ds"
BASE_URL = f"https://huggingface.co/datasets/{DATASET_ID}/resolve/main"

files = list_repo_files(DATASET_ID, repo_type="dataset")

rows = []

for path in files:
    if path.lower().endswith((".jpg", ".png", ".jpeg")):

        label = path.split("/")[0]
        image_name = path.split("/")[-1]

        encoded_path = quote(path)
        image_link = f"{BASE_URL}/{encoded_path}"

        rows.append({
            "image_name": image_name,
            "label": label,
            "image_link": image_link
        })

df = pd.DataFrame(rows)
df.to_csv("train.csv", index=False)

print("train.csv created successfully")


```
